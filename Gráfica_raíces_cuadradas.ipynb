```python
import random
import math
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import numpy as np

# Diccionario de números pitagóricos
numeros_pitagoricos = [
    (5, 1, 2), (8, 2, 2), (10, 1, 3), (13, 2, 3), (17, 1, 4),
    (18, 3, 3), (20, 2, 4), (25, 3, 4), (26, 1, 5), (29, 2, 5),
    (34, 3, 5), (37, 1, 6), (40, 2, 6), (41, 4, 5), (45, 3, 6), (50, 5, 5)
]

estado = {"num": 5, "c1": 1, "c2": 2, "vista": "base"}

# Creación de los botones
boton_nueva = widgets.Button(description='🎲 Generar Nueva Raíz', button_style='success', layout=widgets.Layout(width='220px', height='50px', margin='0 5px 0 0'))
boton_estimacion = widgets.Button(description='📏 Estrategia: Estimación', button_style='info', layout=widgets.Layout(width='220px', height='50px', margin='0 5px 0 0'))
boton_pitagoras = widgets.Button(description='📐 Estrategia: Pitágoras', button_style='warning', layout=widgets.Layout(width='220px', height='50px'))

caja_botones = widgets.HBox([boton_nueva, boton_estimacion, boton_pitagoras], layout=widgets.Layout(justify_content='center', margin='20px 0'))

# Salidas separadas
salida_html = widgets.HTML()
salida_grafico = widgets.Output()

def generar_html():
    n = estado["num"]
    c1 = estado["c1"]
    c2 = estado["c2"]
    vista = estado["vista"]

    html = f"""
    <div style="background-color: #2c3e50; padding: 30px; border-radius: 15px; font-family: Helvetica, Arial, sans-serif; color: white; min-width: 650px; box-shadow: inset 0 0 20px rgba(0,0,0,0.5);">
        <div style="text-align: center; margin-bottom: 30px;">
            <h2 style="color: #bdc3c7; margin: 0; font-size: 24px;">UBICAR EN LA RECTA NUMÉRICA</h2>
            <div style="font-size: 90px; font-weight: bold; color: #f1c40f; margin-top: 20px; display: block; line-height: 1.2; text-shadow: 3px 3px 10px rgba(0,0,0,0.5);">
                &radic;<span style="text-decoration:overline;">&nbsp;{n}&nbsp;</span>
            </div>
        </div>
    """

    if vista == "estimacion":
        raiz_inferior = int(math.sqrt(n))
        raiz_superior = raiz_inferior + 1
        cuad_inf = raiz_inferior**2
        cuad_sup = raiz_superior**2
        decimal = round(math.sqrt(n), 2)

        html += f"""
        <div style="background-color: white; color: #2c3e50; padding: 25px; border-radius: 10px; font-size: 22px; box-shadow: 3px 3px 10px rgba(0,0,0,0.3);">
            <h3 style="color: #3498db; margin-top: 0; border-bottom: 2px solid #3498db; padding-bottom: 10px;">Estrategia 1: Acotación por Cuadrados Perfectos</h3>
            <ul style="line-height: 1.8;">
                <li>Ubicamos los cuadrados perfectos más cercanos a <strong>{n}</strong>:
                    <br><span style="color: #e74c3c; font-weight: bold;">{cuad_inf}</span> &nbsp;&lt;&nbsp; {n} &nbsp;&lt;&nbsp; <span style="color: #e74c3c; font-weight: bold;">{cuad_sup}</span>
                </li>
                <li>Aplicamos la raíz cuadrada a toda la inecuación:
                    <br>&radic;<span style="text-decoration:overline;">{cuad_inf}</span> &nbsp;&lt;&nbsp; &radic;<span style="text-decoration:overline;">{n}</span> &nbsp;&lt;&nbsp; &radic;<span style="text-decoration:overline;">{cuad_sup}</span>
                </li>
                <li>Por lo tanto, la raíz se ubica entre los enteros:
                    <br><span style="font-size: 30px; font-weight: bold; color: #27ae60;">{raiz_inferior} &nbsp;&lt;&nbsp; &radic;<span style="text-decoration:overline;">{n}</span> &nbsp;&lt;&nbsp; {raiz_superior}</span>
                </li>
                <li style="margin-top: 15px; font-size: 18px; color: #7f8c8d;"><em>(Comprobación con calculadora: &approx; {decimal})</em></li>
            </ul>
        </div>
        """

    elif vista == "pitagoras":
        html += f"""
        <div style="background-color: white; color: #2c3e50; padding: 25px; border-radius: 10px; font-size: 22px; box-shadow: 3px 3px 10px rgba(0,0,0,0.3);">
            <h3 style="color: #e67e22; margin-top: 0; border-bottom: 2px solid #e67e22; padding-bottom: 10px;">Estrategia 2: Construcción Geométrica (Pitágoras)</h3>
            <p>Buscamos dos números enteros cuyos cuadrados sumen exactamente <strong>{n}</strong>.</p>

            <div style="background-color: #f8f9fa; padding: 15px; border-radius: 8px; text-align: center; font-size: 32px; margin: 15px 0; border: 1px dashed #bdc3c7;">
                {n} = <span style="color: #c0392b; font-weight: bold;">{c1}<sup>2</sup></span> + <span style="color: #2980b9; font-weight: bold;">{c2}<sup>2</sup></span>
            </div>

            <ul style="line-height: 1.8;">
                <li>Sobre la recta numérica, dibujamos un triángulo rectángulo.</li>
                <li>La base medirá <strong style="color: #c0392b;">{c1} unidades</strong> y la altura <strong style="color: #2980b9;">{c2} unidades</strong>.</li>
                <li>Por el Teorema de Pitágoras (h<sup>2</sup> = a<sup>2</sup> + b<sup>2</sup>), la hipotenusa de ese triángulo medirá exactamente <strong>&radic;<span style="text-decoration:overline;">{n}</span></strong>.</li>
                <li>Finalmente, usamos un compás con centro en 0 para abatir la hipotenusa sobre la recta numérica.</li>
            </ul>
        </div>
        """

    html += "</div>"
    return html

def dibujar_grafico():
    n = estado["num"]
    c1 = estado["c1"]
    c2 = estado["c2"]

    with salida_grafico:
        clear_output(wait=False)

        if estado["vista"] != "pitagoras":
            return

        fig, ax = plt.subplots(figsize=(8, 4.5), dpi=100)

        # Coordenadas del triángulo
        x_tri = [0, c1, c1, 0]
        y_tri = [0, 0, c2, 0]
        ax.plot(x_tri, y_tri, color='#2c3e50', lw=2)
        ax.fill([0, c1, c1], [0, 0, c2], color='#3498db', alpha=0.2)

        # Dibujar el arco del compás
        radio = math.sqrt(n)
        theta = np.linspace(0, math.atan2(c2, c1), 100)
        x_arc = radio * np.cos(theta)
        y_arc = radio * np.sin(theta)
        ax.plot(x_arc, y_arc, linestyle='--', color='#e67e22', lw=2.5)

        # Marcar el punto exacto
        ax.plot(radio, 0, marker='o', markersize=10, color='#e74c3c')
        ax.annotate(f'√{n}', xy=(radio, 0), xytext=(0, 15), textcoords='offset points',
                    ha='center', color='#e74c3c', fontweight='bold', fontsize=14,
                    bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="#e74c3c", lw=1))

        # Etiquetas (AQUÍ ESTÁ EL CAMBIO DE POSICIÓN: de -0.4 a -0.8)
        ax.text(c1 / 2, -0.8, f'{c1}', ha='center', color='#c0392b', fontweight='bold', fontsize=13)
        ax.text(c1 + 0.2, c2 / 2, f'{c2}', va='center', color='#2980b9', fontweight='bold', fontsize=13)

        angulo_grados = math.degrees(math.atan2(c2, c1))
        ax.text(c1 / 2 - 0.2, c2 / 2 + 0.2, f'√{n}', ha='center', va='center',
                rotation=angulo_grados, color='#2c3e50', fontweight='bold', fontsize=13)

        # Configuración del plano (AQUÍ ESTÁ EL CAMBIO DE LÍMITE: de -1 a -1.5)
        ax.set_aspect('equal')
        ax.set_ylim(-1.5, c2 + 1)
        ax.set_xlim(-1, radio + 1)
        ax.axhline(0, color='black', lw=1.5)
        ax.axvline(0, color='black', lw=1.5)
        ax.grid(True, linestyle=':', alpha=0.6)

        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['left'].set_position('zero')
        ax.spines['bottom'].set_position('zero')

        plt.tight_layout()
        plt.show()

def actualizar_pantalla():
    salida_html.value = generar_html()
    dibujar_grafico()

def on_nueva(b):
    seleccion = random.choice(numeros_pitagoricos)
    estado["num"] = seleccion[0]
    estado["c1"] = seleccion[1]
    estado["c2"] = seleccion[2]
    estado["vista"] = "base"
    actualizar_pantalla()

def on_estimacion(b):
    estado["vista"] = "estimacion"
    actualizar_pantalla()

def on_pitagoras(b):
    estado["vista"] = "pitagoras"
    actualizar_pantalla()

# Conectar botones
boton_nueva.on_click(on_nueva)
boton_estimacion.on_click(on_estimacion)
boton_pitagoras.on_click(on_pitagoras)

# Inicializar
on_nueva(None)
caja_principal = widgets.VBox([caja_botones, salida_html, salida_grafico])
display(caja_principal)
```
